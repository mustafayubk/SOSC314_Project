# Week 2 Feasibility Figure

This notebook generates the Week 2 feasibility figure showing the number of comments by video category.

The purpose of this figure is to demonstrate data availability and feasibility, not to perform substantive analysis.

```python
import matplotlib.pyplot as plt

# Comment counts recorded from YouTube (Week 2 pilot)
categories = ['TED-style talk', 'Short film', 'Personal opinion video']
comment_counts = [12816, 3307, 2560]

# Create bar chart
plt.figure()
plt.bar(categories, comment_counts)
plt.xlabel('Video Category')
plt.ylabel('Number of Comments')
plt.title('Number of Comments by Video Category (Pilot Videos)')

plt.tight_layout()
plt.show()
```
