# Week 3: Sampling Plan

## Objective
Construct a manageable but diverse dataset of YouTube comments for Week 3, focusing on data construction, cleaning, and measurement strategy.

## Dataset Design
- Three video categories: TED-style talks, short films, and personal opinion videos
- Three videos per category (9 videos total)
- Up to 500 top-level comments per video (target ~4,500 raw comments)

## Time Stratification
To avoid concentrating videos in a single period, videos are selected from three time bins:
- 2000–2010
- 2010–2020
- Post-2020

Each category contributes one video to each time bin.

## Scope Notes
The dataset size is fixed at this stage to prioritize clarity in data construction, cleaning, and measurement decisions.
